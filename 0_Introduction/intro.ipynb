A collaborative hackathon for exploring carbon-cycle questions with Earth Observation data using [EO-LINCS](https://www.eo-lincs.org/) for carbon-science data access and workflows, and [DeepESDL / Earth System Data Lab](https://earthsystemdatalab.net/) as the working platform.

## What we will work on

Bring a carbon-cycle question, explore suitable data, and develop a small analysis that others can understand and reuse. EO-LINCS provides examples spanning carbon fluxes, forest recovery, model–data relationships, and vegetation-model benchmarking. These are starting points for choosing a question; participants can bring their own ideas. See the [EO-LINCS science cases](https://www.eo-lincs.org/science-cases.html).



## Suggested workflow

1. Read the [platform guide](0_Introduction/Workspace/workspace.md) and confirm your access before the first session.
2. Join the introductions and EO-LINCS tutorial on the 23rd.
3. Choose a question and a small set of relevant variables, locations, and dates.
4. Use the coworking day to explore the data and develop your analysis.
5. Save your notebooks, scripts, and a short presentation
6. Bring results, questions, and feedback to the consolidation session on the 25th.
